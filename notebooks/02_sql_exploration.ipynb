```python
import sqlite3
import pandas as pd

# Chargement du fichier nettoyé
df = pd.read_csv('../data/processed/hotel_bookings_clean.csv')

# Connexion / Création de la base de données dans le dossier sql/
conn = sqlite3.connect('../sql/hotel_analytics.db')

# Transfert des données dans la table 'bookings'
df.to_sql('bookings', conn, if_exists='replace', index=False)
conn.close()
```
